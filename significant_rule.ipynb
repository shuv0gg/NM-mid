```python
#0.0001234 expected out put : 4 

def sign_valu(num):
    num = num.strip()

    if '.' in num:
        # Remove leading zeros
        num = num.lstrip('0')

        integer, decimal = num.split('.')

        # Remove leading zeros in decimal part
        if integer == '':
            decimal = decimal.lstrip('0')

        # Remove trailing zeros only if they are after decimal
        return len(integer + decimal)

    else:
        # Remove leading zeros and trailing zeros
        num = num.lstrip('0').rstrip('0')
        return len(num)


value = input("Enter a value: ")

print("Significant figures:", sign_valu(value))


```

```python
from decimal import Decimal, ROUND_HALF_EVEN
#decimal & banker 
value = input("Enter value: ")
n = int(input("Enter number of significant figures: "))

x = Decimal(value)

# Find the decimal position needed
shift = n - x.adjusted() - 1

quantize_value = Decimal("1e" + str(-shift))

result = x.quantize(quantize_value, rounding=ROUND_HALF_EVEN)

print("Result:", result)
```

```python
value = float(input("Enter value: "))
n = int(input("Enter number of significant figures: "))

# Find the number of digits before the decimal point
temp = abs(value)
digits = 0

while temp >= 1:
    temp = temp / 10
    digits = digits + 1

# Find how many decimal places to keep
decimal_places = n - digits

# Create rounding factor
factor = 1

if decimal_places > 0:
    i = 0
    while i < decimal_places:
        factor = factor * 10
        i = i + 1

    result = int(value * factor + 0.5) / factor

elif decimal_places == 0:
    result = int(value + 0.5)

else:
    i = 0
    while i < -decimal_places:
        factor = factor * 10
        i = i + 1

    result = int(value / factor + 0.5) * factor

print("Result:", result)
```
