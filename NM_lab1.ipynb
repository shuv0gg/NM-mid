```python
import numpy as np
import struct

num = -67.123

# Convert to 32-bit single precision
single = np.float32(num)

# Get 32-bit IEEE 754 representation9
bits = struct.unpack('>I', struct.pack('>f', single))[0]
binary = format(bits, '032b')

# Split into fields
sign = binary[0]
exponent = binary[1:9]
mantissa = binary[9:]

print("Number:", num)
print("Single Precision:", single)
print("32-bit representation:", binary)

print("\nSign     :", sign)
print("Exponent :", exponent)
print("Mantissa :", mantissa)
```

```python
import struct

def float_to_ieee754(num: float) -> str:
    """
    Convert a Python float to IEEE 754 binary representation
    (32-bit single precision).
    """
    try:
        # Pack float into 4 bytes (single precision)
        packed = struct.pack('>f', num)

        # Convert bytes to binary string  4 bytes into a single 32-character binary string.
        bits = ''.join(f'{byte:08b}' for byte in packed)

        return bits

    except Exception as e:
        return f"Error: {e}"


def display_ieee754(num: float):
    """
    Display the IEEE 754 breakdown:
    Sign (1 bit), Exponent (8 bits), Mantissa (23 bits).
    """
    bits = float_to_ieee754(num)

    if bits.startswith("Error"):
        print(bits)
        return

    # Split 32 bits
    sign = bits[0]
    exponent = bits[1:9]
    mantissa = bits[9:]

    print(f"Number: {num}")
    print(f"Binary (32-bit): {bits}")
    print(f"Sign bit: {sign}")
    print(f"Exponent bits: {exponent} (Decimal: {int(exponent, 2)})")
    print(f"Mantissa bits: {mantissa}")

    print("-" * 50)


if __name__ == "__main__":
    try:
        value = float(input("Enter a floating-point number: "))
        display_ieee754(value)

    except ValueError:
        print("Invalid input. Please enter a valid number.")
```
